# 1. Regresión Lineal

¿Recuerdas la ecuación de línea recta de tus clases de matemáticas? ¡Ahí es exactamente donde comienza la regresión lineal! La ecuación básica es maravillosamente simple:

$$y = mx + b$$

En el aprendizaje automático, lo presentamos de forma un poco más formal como:

$$y = \beta_0 + \beta_1 x + \epsilon$$

Donde:

- $y$ es nuestra variable objetivo (lo que estamos tratando de predecir)

- $\beta_0$ es la intersección con el eje y (donde nuestra línea cruza el eje y)

- $\beta_1$ es la pendiente (cuánto cambia y por cada cambio de unidad en x)

- $\epsilon$ es nuestro término de error (¡porque los datos reales no son perfectos!)

## Regresión lineal en acción
Imagina que eres un agente inmobiliario que quiere predecir el precio de una casa basándose en factores como la cantidad de habitaciones y la tasa de criminalidad en la zona. Al utilizar la regresión lineal, puedes desarrollar un modelo que aprenda de datos históricos para realizar predicciones precisas de precios de casas nuevas. Esto te permite no sólo tomar decisiones informadas sino también explicar a tus clientes cómo cada factor contribuye al precio general.

### A considerar:

La regresión lineal es más efectiva cuando existe una relación lineal entre la(s) variable(s) independiente (características) y la variable dependiente (objetivo). Esto significa que el valor objetivo aumenta o disminuye proporcionalmente con el cambio en los valores de entrada.

### Análisis exploratorio de datos

```python
import seaborn as sns
import matplotlib.pyplot as plt

sns.scatterplot(x=df['MedInc'], y=df['MedHouseVal'])
plt.xlabel("Ingreso medio")
plt.ylabel("Valor medio de vivienda")
plt.title("Relación entre ingreso y precio")
plt.show()
```

```python
def limpiar_outliers(dataset, columnas):
    df_limpio = dataset.copy()
    for col in columnas:
        Q1 = df_limpio[col].quantile(0.25)
        Q3 = df_limpio[col].quantile(0.75)
        IQR = Q3 - Q1

        limite_inferior = Q1 - 1.5 * IQR
        limite_superior = Q3 + 1.5 * IQR

        # Filtrar el DataFrame
        df_limpio = df_limpio[(df_limpio[col] >= limite_inferior) & (df_limpio[col] <= limite_superior)]

    return df_limpio

# Columnas donde sospechamos que hay valores locos
cols_a_limpiar = ['MedInc', 'AveRooms', 'AveBedrms', 'Population', 'AveOccup']

df_final = limpiar_outliers(df, cols_a_limpiar)

print(f"Registros originales: {len(df)}")
print(f"Registros tras limpieza: {len(df_final)}")
# Calcular la matriz de correlación
corr_matrix = df_final.corr()

# Visualizar con un heatmap
plt.figure(figsize=(10, 8))
sns.heatmap(corr_matrix, annot=True, cmap='coolwarm', fmt=".2f")
plt.title("Matriz de Correlación de California Housing")
plt.show()
```

Realizamos una limpieza de outliners y aplicamos una matriz de correlacion, recordar que las columnas que tenemos son:

AveOccup = Promedio de habitantes en el bloque

MedHouseVal = Media de valor de la casa

# 2. Regresión Logística

La regresión logística funciona mejor para tareas de clasificación binaria, donde el resultado sólo puede pertenecer a una de dos categorías (p. ej., Sí/No, 0/1). El modelo funciona bien cuando los puntos de datos se pueden separar mediante un límite lineal. Los ejemplos incluyen:

- Clasificar correos electrónicos como spam o no spam.

- Determinar si un cliente comprará un producto o no.

- Predecir si un estudiante aprobará o reprobará en función de sus horas de estudio.

Si una simple línea divisoria puede separar eficazmente las dos categorías, la regresión logística es una opción adecuada y poderosa para realizar estas clasificaciones.


La regresión logística utiliza la función sigmoide para modelar probabilidades:

$$\sigma(z) = \frac{1}{1 + e^{-z}}$$

La belleza de esta función radica en sus propiedades:

- Exprime cualquier entrada en un rango entre 0 y 1

- Cuando $z = 0, \sigma (z) = 0,5$

- Cuando $z$ tiende a ∞, $\sigma (z)$ tiende a 1

- Cuando $z$ tiende a -∞, $\sigma (z)$ tiende a 0

### Matriz de confusión

La matriz de confusión es una tabla utilizada para describir el rendimiento de un modelo de clasificación. Esta muestra el número de predicciones de verdaderos positivos (TP), verdaderos negativos (TN), falsos positivos (FP) y falsos negativos (FN).

|                      | Predicho positivo | Negativo previsto |
|----------------------|-------------------|-------------------|
| **Positivo real**    | Verdadero positivo (TP) | Falso Negativo (FN) |
| **Negativo real**    | Falso positivo (FP)     | Verdadero Negativo (TN) |

- Verdadero positivo (TP): El modelo predijo correctamente una clase positiva.

- Verdadero Negativo (TN): El modelo predijo correctamente una clase negativa.

- Falso positivo (FP) (Error tipo I): El modelo predijo incorrectamente una clase positiva (una "falsa alarma").

- Falso Negativo (FN) (Error tipo II): el modelo predijo incorrectamente una clase negativa (pasó por alto un caso positivo).

### Informe de clasificación

El informe de clasificación proporciona un desglose detallado del rendimiento del modelo, incluyendo **Precisión**, **Recordar** y **Puntuación F1** para cada clase.

---

#### Métricas clave en el informe de clasificación

##### Precisión (Valor Predictivo Positivo)
- **Definición:** La proporción de predicciones positivas que realmente fueron correctas.  
$$\text {Precisión} = \frac{TP} {TP + FP}$$
- **Interpretación:**  
  - Alta precisión → pocos falsos positivos.  
  - Útil cuando el costo de un falso positivo es alto (ejemplo: predecir una enfermedad cuando no está presente).

---

##### Recordatorio (Sensibilidad o Tasa Verdadera Positiva)
- **Definición:** La proporción de positivos reales que se predijeron correctamente.  
$$\text {Recall} = \frac{TP} {TP + FN}
$$
- **Interpretación:**  
  - Alto recuerdo → pocos falsos negativos.  
  - Útil cuando el costo de un falso negativo es alto (ejemplo: pasar por alto un diagnóstico de enfermedad).

---

##### Puntuación F1
- **Definición:** La media armónica de precisión y recuperación.  
$$\text{F1-Score} = 2 \times \frac{\text{Precision} \times \text{Recall}}{\text{Precision} + \text{Recall}}$$
- **Interpretación:**  
  - Equilibra ambas métricas y es útil cuando necesitas una única puntuación de rendimiento.  
  - Una puntuación F1 perfecta es **1.0**, lo que indica precisión y recuperación perfectas.
